```python
metadata_data = [
    ("weather_raw", "station_id", "Unique identifier for weather station"),
    ("weather_raw", "city", "City where the station is located"),
    ("weather_raw", "temperature_c", "Temperature measured in Celsius"),
    ("weather_raw", "weather_condition", "Observed weather condition"),
    ("station_info", "station_id", "Unique identifier for weather station"),
    ("station_info", "region", "Geographical region code"),
    ("station_info", "station_type", "Classification of weather station")
]

columns = ["table_name", "column_name", "description"]

metadata_df = spark.createDataFrame(metadata_data, columns)
metadata_df.show(truncate=False)

```

```python
metadata_df.write.mode("overwrite").saveAsTable("metadata_definitions")
```

```python
spark.sql("SELECT * FROM metadata_definitions").show(truncate=False)
```

```python
weather_data = [
    (1, "New York", 22.5, "Rain"),
    (2, "Dallas", 30.1, "Clear"),
    (3, "Chicago", 18.9, "Cloudy")
]

weather_columns = [
    "station_id",
    "city",
    "temperature_c",
    "weather_condition"
]

weather_df = spark.createDataFrame(weather_data, weather_columns)

weather_df.write.mode("overwrite").saveAsTable("weather_raw")
```

```python
station_data = [
    (1, "US-NY", "Urban"),
    (2, "US-TX", "Urban"),
    (3, "US-IL", "Urban")
]

station_columns = [
    "station_id",
    "region",
    "station_type"
]

station_df = spark.createDataFrame(station_data, station_columns)

station_df.write.mode("overwrite").saveAsTable("station_info")
```

```python
tables = ["weather_raw", "station_info"]

metadata = []

for table in tables:
    cols = spark.sql(f"DESCRIBE TABLE {table}").collect()
    
    for col in cols:
        if col.col_name and not col.col_name.startswith("#"):
            metadata.append((table, col.col_name))
```

```python
metadata
```

```python
actual_metadata_df = spark.createDataFrame(
    metadata,
    ["table_name", "column_name"]
)

actual_metadata_df.show()
```

```python
actual_metadata_df.write.mode("overwrite").saveAsTable("actual_metadata")
```

```python
spark.sql("SELECT * FROM actual_metadata").show(truncate=False)
```
